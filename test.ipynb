```python
from dotenv import load_dotenv
load_dotenv()
from datasets import load_dataset

ds = load_dataset("hotpotqa/hotpot_qa", "distractor")
```

```python
data = ds["validation"]

samples = data.select(range(15))

print(samples[0])
```

```python
print(samples[0]["context"]['sentences'][2])
```

```python
import json

pilot = []

for item in samples:
    pilot.append(item)


with open(
    "data/pilot_questions.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        pilot,
        f,
        indent=2,
        ensure_ascii=False
    )
```
